## La demande client

Tu reprends le client **Acerox** (cas M3). Avant d'ingérer des **logs textuels** dans la base, leur DPO exige de **repérer et masquer les données personnelles** (noms de personnes, lieux, organisations) — une étape de **pseudonymisation**.

Ta mission : à partir d'un modèle de **reconnaissance d'entités nommées (NER)** pré-entraîné, écris un mini-pipeline qui, pour une ligne de log, **détecte les entités** (personnes / lieux / organisations) et renvoie le texte avec ces entités **remplacées** par un marqueur (ex. `[PER]`, `[LOC]`, `[ORG]`).

Exemple attendu :

```
Entrée : "Marie Dupont a signalé une panne sur le site de Bordeaux (fournisseur Acerox)."
Sortie : "[PER] a signalé une panne sur le site de [LOC] (fournisseur [ORG])."
```

## Ce que tu dois produire (sans étapes guidées)

1. **Choisir** un modèle NER adapté au **français** (pense à la checklist d'adoption : tâche, langue, licence, taille). 
   *Piste validée et robuste (pas de dépendance exotique) : `Babelscape/wikineural-multilingual-ner`, chargé en `pipeline("token-classification", aggregation_strategy="simple")`.*
2. **Charger** le modèle et l'appliquer à quelques lignes de log que tu inventes.
3. **Écrire** la fonction de pseudonymisation (remplacer chaque entité détectée par son marqueur `[PER]/[LOC]/[ORG]`).
4. **Sobriété** : mesure la taille du modèle et la latence par ligne. Demande-toi : *un simple jeu d'expressions régulières suffirait-il pour mon format de logs, ou la variabilité justifie-t-elle un modèle NER ?*

> 💡 **Lien parcours** : c'est exactement le geste de **pseudonymisation** vu en M2-B2 / M3 (RGPD), mais cette fois avec un **modèle pré-entraîné** plutôt qu'à la main. Et l'arbitrage *regex vs modèle* est un vrai choix de sobriété.
